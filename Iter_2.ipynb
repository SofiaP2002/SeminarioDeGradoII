```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

variables = ['reporter_id', 'assignee_id', 'priority', 'project_ID']
target = 'target'

# Asegurar que existan en el df
variables_validas = [col for col in variables if col in df.columns]

resultados_cramer = []

for col in variables_validas:
  # Imputar nulos con una categoría constante para mantener la consistencia del análisis anterior
  temp_series = df[col].fillna('No especificado/Nulo').astype(str)

  # Crear tabla de contingencia
  tabla_contingencia = pd.crosstab(temp_series, df[target])

  # Calcular Chi-cuadrado
  chi2, p_val, dof, expected = chi2_contingency(tabla_contingencia)

  # Calcular V de Cramér
  n = tabla_contingencia.sum().sum()
  r, c = tabla_contingencia.shape
  v_cramer = np.sqrt(chi2 / (n * min(r - 1, c - 1)))

  resultados_cramer.append({
            'Variable': col,
            'Chi2 Stat': chi2,
            'p-valor': p_val,
            'V de Cramér': v_cramer
  })

# Crear la tabla y ordenarla una sola vez fuera del bucle
df_cramer = pd.DataFrame(resultados_cramer).sort_values(by='V de Cramér', ascending=False)

print("--- ASOCIACIÓN CON V DE CRAMÉR RESPECTO AL TARGET ---")
display(df_cramer)
```

### 7. Manejo de Outliers en 'Story_Point_Original' (Z-score)


```python
import numpy as np

# Eliminar Story Points negativos
df = df[df['story_point_inicial'] >= 0].copy()

# Z-score por proyecto
z_score_threshold = 3

media = df.groupby('Project_ID')['story_point_inicial'].transform('mean')
std = df.groupby('Project_ID')['story_point_inicial'].transform('std')

std = std.replace(0, np.nan)

z_scores = (
    (df['story_point_inicial'] - media) / std
)

mask_outlier = z_scores.abs() > z_score_threshold
mask_outlier = mask_outlier.fillna(False)

print(f"Outliers eliminados: {mask_outlier.sum()}")

df = df.loc[~mask_outlier].copy()

print("Forma final:", df.shape)
```

### 7. Manejo de Outliers completando con la mediana


```python
import numpy as np
import pandas as pd

if 'df' in locals() and df is not None:
    print(f"Estado inicial del DataFrame: {df.shape[0]} filas.")
    print(f"Nulos iniciales en SP: {df['story_point_inicial'].isnull().sum()}")

    # 1. Convertir valores negativos a NaN
    negativos_mask = df['story_point_inicial'] < 0
    num_negativos = negativos_mask.sum()
    df.loc[negativos_mask, 'story_point_inicial'] = np.nan
    print(f"Valores negativos convertidos a NaN: {num_negativos}")

    # 2. Identificar y reemplazar outliers por proyecto usando Z-score robusto o IQR
    # Usaremos el método IQR (Rango Intercuartílico) por cada Project_ID para identificar outliers
    def replace_outliers_with_nan(group):
        q1 = group.quantile(0.25)
        q3 = group.quantile(0.75)
        iqr = q3 - q1
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr
        # Marcar como NaN los valores fuera de los límites (solo si no son nulos de por sí)
        is_outlier = (group < lower_bound) | (group > upper_bound)
        group[is_outlier] = np.nan
        return group

    # Aplicar la función de outliers por proyecto
    sp_antes_outliers_nulos = df['story_point_inicial'].isnull().sum()
    df['story_point_inicial'] = df.groupby('Project_ID')['story_point_inicial'].transform(replace_outliers_with_nan)
    sp_despues_outliers_nulos = df['story_point_inicial'].isnull().sum()
    print(f"Outliers detectados por proyecto y convertidos a NaN: {sp_despues_outliers_nulos - sp_antes_outliers_nulos}")

    # 3. Imputar todos los NaN (faltantes, negativos y outliers) con la mediana de su respectivo Project_ID
    df['story_point_inicial'] = df.groupby('Project_ID')['story_point_inicial'].transform(lambda x: x.fillna(x.median()))

    # Si aún quedasen nulos debido a proyectos sin ningún SP válido, imputar con la mediana global
    global_median = df['story_point_inicial'].median()
    nulos_restantes = df['story_point_inicial'].isnull().sum()
    if nulos_restantes > 0:
        df['story_point_inicial'] = df['story_point_inicial'].fillna(global_median)
        print(f"Se imputaron {nulos_restantes} valores restantes usando la mediana global ({global_median})")

    print(f"Nulos finales en la columna 'story_point_inicial': {df['story_point_inicial'].isnull().sum()}")
    display(df['story_point_inicial'].describe())
else:
    print("El DataFrame 'df' no está disponible en el entorno.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
import warnings
warnings.filterwarnings('ignore') # Suprimir advertencias

# --- AJUSTA ESTE UMBRAL MANUALMENTE POR CÓDIGO ---
umbral_manual = 0.40
# -------------------------------------------------

# Asumiendo que X_train, X_test, y_train, y_test y la función mostrar_resultados
# han sido definidos en celdas anteriores, como en el código previamente adaptado.

# 1. Función para crear y entrenar el modelo con class_weight='balanced'
def run_model_balanced(X_train_processed, y_train_data):
    clf = LogisticRegression(
        C=1.0,
        penalty='l2',
        random_state=1,
        solver="saga", # Usar 'saga' para consistencia y soporte de 'balanced' class_weight
        max_iter=2000, # Aumentar max_iter para asegurar convergencia
        class_weight="balanced"
    )
    clf.fit(X_train_processed, y_train_data)
    return clf

# 2. Ejecutar el modelo con class_weight='balanced'
print("\n--- Entrenando Modelo con class_weight='balanced' ---")
model_balanced = run_model_balanced(X_train, y_train)
print("Modelo de Regresión Logística (balanced) entrenado.")

# 3. Realizar predicciones probabilísticas
pred_proba_balanced = model_balanced.predict_proba(X_test)[:, 1]

# 4. Clasificar aplicando el umbral configurado manual por código
pred_y_balanced_custom = (pred_proba_balanced >= umbral_manual).astype(int)

print(f"\n--- Métricas de Evaluación en el Conjunto de Prueba (balanced - Umbral Manual: {umbral_manual}) ---")

# Calcular Matriz de Confusión para derivar exactitudes
tn, fp, fn, tp = confusion_matrix(y_test, pred_y_balanced_custom).ravel()

specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0

# Exactitud por clase (Accuracy per class)
accuracy_clase_0 = tn / (tn + fp) if (tn + fp) > 0 else 0
accuracy_clase_1 = tp / (tp + fn) if (tp + fn) > 0 else 0

# Exactitud General (Overall Accuracy)
exactitud_general = (tn + tp) / (tn + fp + fn + tp) if (tn + fp + fn + tp) > 0 else 0

print(f"Especificidad (Specificity): {specificity:.4f}")
print(f"Sensibilidad (Recall Clase 1): {sensitivity:.4f}")
print("\n--- EXACTITUD Y RENDIMIENTO DE CLASIFICACIÓN ---")
print(f"Exactitud Clase 0 (No Cambió / Negativo): {accuracy_clase_0 * 100:.2f}%")
print(f"Exactitud Clase 1 (Sí Cambió / Positivo): {accuracy_clase_1 * 100:.2f}%")
print(f"EXACTITUD GENERAL (Overall Accuracy): {exactitud_general * 100:.2f}%")
print("-----------------------------------------------")

mostrar_resultados(y_test, pred_y_balanced_custom, pred_proba_balanced)
```

```python
# ============================================================
# 0. IMPORTACIONES REQUERIDAS
# ============================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    balanced_accuracy_score,
    average_precision_score,
    roc_auc_score
)

# ============================================================
# 1. DEFINIR VARIABLE OBJETIVO
# ============================================================

y = df['target'].astype(int)

df_model = df.copy()

df_model['Texto_Combinado'] = (
    df_model['title'].fillna('') + ' ' +
    df_model['description'].fillna('')
)

df_model['reporter_id'] = (
    df_model['reporter_id']
    .fillna('No Reporter')
    .astype(str)
)

df_model['experiencia_en_proyecto'] = (
    df_model['experiencia_en_proyecto'].fillna(0)
)

# ============================================================
# 2. DEFINIR CARACTERÍSTICAS
# ============================================================

numerical_features = [
    'story_point_inicial',
    'experiencia_en_proyecto'
]

categorical_features = [
    'reporter_id'
]

text_feature_col = 'Texto_Combinado'
components_text_col = 'First_Component'

features_to_use = (
    numerical_features +
    categorical_features +
    [text_feature_col, components_text_col]
)

X = df_model[features_to_use]


# ============================================================
# 3. PRIMERA DIVISIÓN: TRAIN (80%) Y TEST (20%)
# ============================================================

X_train_cal_raw, X_test_raw, y_train_cal, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ============================================================
# 4. SEGUNDA DIVISIÓN: TRAIN (75%) Y CALIBRACIÓN (25%)
# ============================================================

X_train_raw, X_cal_raw, y_train, y_cal = train_test_split(
    X_train_cal_raw,
    y_train_cal,
    test_size=0.25,
    random_state=42,
    stratify=y_train_cal
)


# ============================================================
# 5. VERIFICAR DISTRIBUCIÓN DE LOS CONJUNTOS
# ============================================================

print("\n--- DIMENSIONES DE LOS CONJUNTOS ---")

print(f"X_train: {X_train_raw.shape}")
print(f"X_cal:   {X_cal_raw.shape}")
print(f"X_test:  {X_test_raw.shape}")

print(f"\ny_train: {y_train.shape}")
print(f"y_cal:   {y_cal.shape}")
print(f"y_test:  {y_test.shape}")


print("\n--- DISTRIBUCIÓN DE CLASES ---")

distribucion = pd.DataFrame({
    'Train': y_train.value_counts(normalize=True),
    'Calibración': y_cal.value_counts(normalize=True),
    'Test': y_test.value_counts(normalize=True)
}).fillna(0)

print((distribucion * 100).round(2))


# ============================================================
# 6. PREPROCESAMIENTO
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            StandardScaler(),
            numerical_features
        ),
        (
            'cat',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=True
            ),
            categorical_features
        ),
        (
            'text_tfidf_desc_title',
            TfidfVectorizer(
                lowercase=True,
                strip_accents='unicode',
                sublinear_tf=True,
                min_df=3,
                max_df=0.95,
                ngram_range=(1, 2),
                max_features=10000
            ),
            text_feature_col
        ),
        (
            'text_tfidf_components',
            TfidfVectorizer(
                lowercase=True,
                strip_accents='unicode',
                sublinear_tf=True,
                min_df=1,
                max_df=0.95,
                ngram_range=(1, 1),
                max_features=1000
            ),
            components_text_col
        )
    ],
    remainder='passthrough',
    verbose_feature_names_out=False
)


# ============================================================
# 7. AJUSTAR PREPROCESAMIENTO SOLO CON TRAIN
# ============================================================

X_train = preprocessor.fit_transform(X_train_raw)

X_cal = preprocessor.transform(X_cal_raw)

X_test = preprocessor.transform(X_test_raw)


# ============================================================
# 8. VERIFICAR DIMENSIONES FINALES
# ============================================================

print("\n--- DIMENSIONES TRAS EL PREPROCESAMIENTO ---")

print(f"X_train: {X_train.shape}")
print(f"X_cal:   {X_cal.shape}")
print(f"X_test:  {X_test.shape}")


# ============================================================
# 9. FUNCIÓN DE EVALUACIÓN CON EXACTITUD POR CLASE
# ============================================================

def mostrar_resultados(y_true, y_pred_values, y_pred_proba=None):

    LABELS = sorted(y_true.unique().astype(str))
    conf_matrix = confusion_matrix(y_true, y_pred_values)

    plt.figure(figsize=(12, 12))
    ax = sns.heatmap(
        conf_matrix,
        xticklabels=LABELS,
        yticklabels=LABELS,
        annot=True,
        fmt="d",
        annot_kws={"size": 20}
    )

    ax.tick_params(axis='x', labelsize=18)
    ax.tick_params(axis='y', labelsize=18)

    plt.title("Confusion matrix")
    plt.ylabel("True class", fontsize=22)
    plt.xlabel("Predicted class", fontsize=22)
    plt.show()

    print(classification_report(y_true, y_pred_values))

    # Exactitud por clase (Recall para cada clase binaria)
    tn, fp, fn, tp = conf_matrix.ravel()
    accuracy_clase_0 = tn / (tn + fp) if (tn + fp) > 0 else 0
    accuracy_clase_1 = tp / (tp + fn) if (tp + fn) > 0 else 0

    print("--- EXACTITUD POR CLASE (ACCURACY PER CLASS) ---")
    print(f"Exactitud Clase 0 (No Cambió / Negativo): {accuracy_clase_0 * 100:.2f}%")
    print(f"Exactitud Clase 1 (Sí Cambió / Positivo): {accuracy_clase_1 * 100:.2f}%")
    print("-----------------------------------------------")

    balanced_acc = balanced_accuracy_score(
        y_true,
        y_pred_values
    )
    print(f"Balanced Accuracy: {balanced_acc:.4f}")

    if y_pred_proba is not None:
        avg_precision = average_precision_score(
            y_true,
            y_pred_proba
        )
        roc_auc = roc_auc_score(
            y_true,
            y_pred_proba
        )

        print(f"Average Precision (PR-AUC): {avg_precision:.4f}")
        print(f"ROC AUC: {roc_auc:.4f}")

        print("\nPrimeras 10 predicciones:")
        sample_df = pd.DataFrame({
            'Real': y_true.reset_index(drop=True).head(10),
            'Predicha': pd.Series(y_pred_values).head(10),
            'Probabilidad': pd.Series(y_pred_proba).head(10)
        })
        display(sample_df)
```

```python
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    f1_score,
    precision_score,
    recall_score
)
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

umbral_manual = 0.13

predicciones_rf = {
    'RF Original': (
        pred_proba_rf_original >= umbral_manual
    ).astype(int),

    'RF Sigmoid': (
        pred_proba_rf_sigmoid >= umbral_manual
    ).astype(int),

    'RF Isotonic': (
        pred_proba_rf_isotonic >= umbral_manual
    ).astype(int)
}

resultados_rf = []

fig, axes = plt.subplots(1, 3, figsize=(19, 6))

for ax, (nombre, pred) in zip(
    axes,
    predicciones_rf.items()
):
    # Matriz de confusión
    cm = confusion_matrix(
        y_test,
        pred,
        labels=[0, 1]
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues',
        xticklabels=['Predicho 0', 'Predicho 1'],
        yticklabels=['Real 0', 'Real 1'],
        annot_kws={'size': 16},
        ax=ax
    )

    ax.set_title(f'{nombre}\nUmbral: {umbral_manual}')
    ax.set_xlabel('Predicción')
    ax.set_ylabel('Valor real')

    # Calcular soporte para cada clase
    support_0 = np.sum(y_test == 0)
    support_1 = np.sum(y_test == 1)

    # Métricas por clase
    for clase in [0, 1]:
        precision = precision_score(
            y_test,
            pred,
            labels=[clase],
            average=None,
            zero_division=0
        )[0]

        recall = recall_score(
            y_test,
            pred,
            labels=[clase],
            average=None,
            zero_division=0
        )[0]

        f1 = f1_score(
            y_test,
            pred,
            labels=[clase],
            average=None,
            zero_division=0
        )[0]

        support = support_0 if clase == 0 else support_1

        resultados_rf.append({
            'Modelo': nombre,
            'Clase': clase,
            'precision': precision,
            'recall': recall,
            'f1-score': f1,
            'support': support
        })

plt.suptitle(
    'Random Forest - Matrices de confusión',
    fontsize=16
)
plt.tight_layout()
plt.show()

# ============================================================
# CREACIÓN DE TABLAS DE METRICAS POR MODELO EN EL FORMATO SOLICITADO
# ============================================================
df_metricas_rf = pd.DataFrame(resultados_rf)

modelos_disponibles = df_metricas_rf['Modelo'].unique()

for mod in modelos_disponibles:
    print(f"\n--- TABLA DE MÉTRICAS: {mod.upper()} ---")
    df_modelo_filtro = df_metricas_rf[df_metricas_rf['Modelo'] == mod][['Clase', 'precision', 'recall', 'f1-score', 'support']].set_index('Clase')
    # Convertir el soporte a entero para que se muestre correctamente sin decimales
    df_modelo_filtro['support'] = df_modelo_filtro['support'].astype(int)
    display(df_modelo_filtro.round(4))
```
